```python
import pandas as pd
import numpy as np

data = pd.read_excel(r"C:\Users\HP\OneDrive\Desktop\Delinquency\Delinquency_Cleaned_Data.xlsx")
data.head()
```

```python
data.info()
```

```python
total_month = 6
data['Delinquency_Rate'] = data['Missed_Payments'] / total_month
```

```python
data.head()
```

```python
def delinquency_band(rate):
    if rate <= 0.16:
        return 'Low'
    elif rate<= 0.33:
        return 'Medium'
    else:
        return 'HIgh'
data['Delinquency_Risk_Band'] = data['Delinquency_Rate'].apply(delinquency_band)
```

```python
data['Delinquency_Risk_Band'].tail()
```

```python
# High Credit Utilization Flag
data['High_Utilization_Flag'] = (data['Credit_Utilization'] > 0.30).astype(int)
```

```python
# High DTI Flag
data['High_DTI_Flag'] = (data['Debt_to_Income_Ratio'] > 0.40).astype(int)
data.head()
```

```python
data.info()
```

```python
def loan_size_bucket(balance):
    if balance <= 200000:
        return 'Small'
    elif balance <= 500000:
        return 'Medium'
    else:
        return 'Large'

data['Loan_Size'] = data['Loan_Balance'].apply(loan_size_bucket)
```
